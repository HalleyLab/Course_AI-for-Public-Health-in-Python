```python
#@title Advanced LLM Text Generator

import torch
from transformers import pipeline

model = "EleutherAI/gpt-neo-125m" #@param ["gpt2", "EleutherAI/gpt-neo-125m", "EleutherAI/gpt-neo-1.3B"]
prompt = "Tomorrow is the last day of the humankind" #@param {type:"string"}

#@markdown ---
#@markdown ### **Parameters**
max_length = 160 #@param {type:"slider", min:10, max:300, step:10}
temperature = 1.2 #@param {type:"slider", min:0.1, max:2.0, step:0.1}
top_k = 50 #@param {type:"slider", min:0, max:100, step:1}
top_p = 0.7 #@param {type:"slider", min:0.1, max:1.0, step:0.05}
num_return_sequences = 3 #@param {type:"slider", min:1, max:5, step:1}

device = 0 if torch.cuda.is_available() else -1
generator = pipeline("text-generation", model=model, device=device)

output = generator(
    prompt,
    max_length=max_length,
    temperature=temperature,
    top_k=top_k,
    top_p=top_p,
    num_return_sequences=num_return_sequences,
    do_sample=True,
    truncation=True,
    pad_token_id=generator.tokenizer.eos_token_id
)

for idx, res in enumerate(output):
    print(f"\n *** Result {idx + 1}")
    print()
    print(res['generated_text'])
```
